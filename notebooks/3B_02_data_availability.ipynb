```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import yfinance as yf

asset_universe = {
    "Equities" : ["SPY", "VGK", "EWJ", "EEM"],
    "Commodities" : ["USO", "DBA", "DBB", "DBC"],
    "Bonds" : ["IEF", "TLT", "LQD", "HYG"],
    "Forex" : ["FXE", "FXY", "FXB", "FXA"]
}

tickers = []

for assets in asset_universe.values():
    tickers.extend(assets)

# print(tickers)

def download_prices(tickers, start_date, end_date, auto_adjust=True, progress=False):
    prices = yf.download(tickers, start=start_date, end=end_date, auto_adjust=auto_adjust, progress=progress)["Close"]
    return prices

prices = download_prices(tickers, "2008-01-01", "2026-01-01")
print(prices.head())
print(prices.tail())
print(prices.isna().sum())
print(prices.index.is_monotonic_increasing)
print(prices.shape)
print(prices.index.duplicated().sum())
print((prices <= 0).sum())
print((prices > 0).sum() - prices.shape[0])

monthly_prices = prices.resample("ME").last()
print(monthly_prices.head())
print(monthly_prices.tail())
print(monthly_prices.shape)
print(monthly_prices.index.min())
print(monthly_prices.index.max())
print(monthly_prices.isna().sum())

momentum_12m = monthly_prices.pct_change(periods=12)
print(momentum_12m.loc["2009-12-31"])
```
